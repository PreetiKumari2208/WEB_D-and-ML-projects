```python
from langchain.prompts import PromptTemplate
template_name=PromptTemplate(
    input_variables=['country'],
    template="I want to open a {country} restaurant ,suggest name for this."

)
template_name.format(country="Indian")
```

```python
from langchain.chains import LLMChain
chain=LLMChain(llm=llm,prompt=template_name)
chain.run("Chinese")
```

simpleSequential chain gives only final output

```python

template_name_1=PromptTemplate(
    input_variables=['subject'],
    template="Suggest a interesting book of {subject} with author name for engineering student."

)
chain_1=LLMChain(llm=llm,prompt=template_name_1)

template_name_2=PromptTemplate(
    input_variables=['book'],
    template="Suggest imporatnt topics of this {book} in the list format for exam preparation."

)
chain_2=LLMChain(llm=llm,prompt=template_name_2)
```

```python
from langchain.chains import SimpleSequentialChain
chain=SimpleSequentialChain(chains=[chain_1,chain_2])
chain.run("operating system")
```

squentialChain gives intermidiate outputs also

```python
template_name_1=PromptTemplate(
    input_variables=['subject'],
    template="Suggest a interesting book of {subject} with author name for engineering student."

)
chain_1=LLMChain(llm=llm,prompt=template_name_1,output_key="book")

template_name_2=PromptTemplate(
    input_variables=['book'],
    template="Suggest imporatnt topics of this {book} in the list format for exam preparation."

)
chain_2=LLMChain(llm=llm,prompt=template_name_2,output_key="important_topics")
```

```python
from langchain.chains import SequentialChain
chain=SequentialChain(chains=[chain_1,chain_2],
                      input_variables=["subject"],
                      output_variables=["book","important_topics"]
                      )

chain({'subject':'operating Sysrem'})
```
